```python
import pandas as pd

df = pd.read_csv("smart_traffic_management_dataset.csv")
df.head()
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
df["hour"] = df["timestamp"].dt.hour
df = df.dropna()
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
df["hour"] = df["timestamp"].dt.hour
print(df.head())
```

```python
hourly = df.groupby("hour")["traffic_volume"].mean().reset_index()
location = df.groupby("location_id")["traffic_volume"].mean().reset_index()
weather = df.groupby("weather_condition")["traffic_volume"].mean().reset_index()
accident = df["accident_reported"].value_counts().reset_index()
accident.columns = ["accident", "count"]
```

```python
import os

os.makedirs("output", exist_ok=True)
hourly.to_csv("output/hourly.csv", index=False)
location.to_csv("output/location.csv", index=False)
weather.to_csv("output/weather.csv", index=False)
accident.to_csv("output/accident.csv", index=False)
```
